### Sentiment Analysis on IMDB Dataset
We will build a deep learning model to classify movie reviews as positive or negative using the IMDB dataset.

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential

# Set parameters
max_features = 10000  # Number of words to consider as features
maxlen = 500          # Cut texts after this number of words
batch_size = 32

# Load the dataset
print('Loading data...')
(input_train, y_train), (input_test, y_test) = imdb.load_data(num_words=max_features)
print(len(input_train), 'train sequences')
print(len(input_test), 'test sequences')
```

```python
print('Pad sequences (samples x time)')
input_train = pad_sequences(input_train, maxlen=maxlen)
input_test = pad_sequences(input_test, maxlen=maxlen)
print('input_train shape:', input_train.shape)
print('input_test shape:', input_test.shape)
```

```python
from tensorflow.keras.layers import Embedding, LSTM, Dense

# Define the model architecture
model = Sequential([
    Embedding(max_features, 32, input_length=maxlen),
    LSTM(32),
    Dense(1, activation='sigmoid')
])

model.compile(optimizer='rmsprop',
              loss='binary_crossentropy',
              metrics=['acc'])

model.summary()
```

```python
# Train the model
history = model.fit(input_train, y_train,
                    epochs=10,
                    batch_size=batch_size,
                    validation_split=0.2)
```

```python
# Evaluate model performance
loss, accuracy = model.evaluate(input_test, y_test)
print(f'Test accuracy: {accuracy:.4f}')
```

### Test the Model with Your Own Movie Review
Use the input field below to type a review and see whether the model classifies it as **Positive** or **Negative**.

```python
import ipywidgets as widgets
import re
from tensorflow.keras.datasets import imdb

# Retrieve the word index to map text reviews to integer sequences
word_index = imdb.get_word_index()

def preprocess_review(text):
    # Remove punctuation and convert to lowercase
    cleaned_text = re.sub(r'[^\w\s]', '', text.lower())
    words = cleaned_text.split()

    # Convert words to indices based on the IMDB dataset index conventions
    # 0 = padding, 1 = start, 2 = unknown
    sequence = [1] # Start token
    for word in words:
        if word in word_index:
            val = word_index[word] + 3 # Shift indices by 3
            if val < max_features:
                sequence.append(val)
            else:
                sequence.append(2)
        else:
            sequence.append(2)

    # Pad the sequence to match maxlen
    padded_sequence = pad_sequences([sequence], maxlen=maxlen)
    return padded_sequence

# Create interactive widgets
text_input = widgets.Textarea(
    value='This movie was absolutely amazing! The acting was great and the plot kept me engaged.',
    placeholder='Type your review here...',
    description='Review:',
    layout=widgets.Layout(width='80%', height='100px')
)

button = widgets.Button(description="Analyze Sentiment", button_style='primary')
output = widgets.Output()

def on_button_clicked(b):
    with output:
        output.clear_output()
        review_text = text_input.value
        if not review_text.strip():
            print("Please enter a review to analyze.")
            return

        # Process and predict
        processed_input = preprocess_review(review_text)
        prediction = model.predict(processed_input, verbose=0)[0][0]

        # Output result
        sentiment = "Positive 😊" if prediction >= 0.5 else "Negative 😞"
        print(f"Review: \"{review_text}\"")
        print(f"Confidence Score: {prediction:.4f}")
        print(f"Predicted Sentiment: {sentiment}")

button.on_click(on_button_clicked)
display(text_input, button, output)
```
